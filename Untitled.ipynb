```python
# Data manipulation and numerical operations
import pandas as pd
import numpy as np

# Data visualization libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Preprocessing and train/test splitting
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Machine Learning Classifiers
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC

# Evaluation Metrics (matching the 5 core metrics from your reference)
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    recall_score,
    precision_score,
    confusion_matrix,
    classification_report
)


# Ignore minor warnings for a cleaner notebook presentation
import warnings
warnings.filterwarnings('ignore')

# Set visual style for all plots
sns.set_theme(style="whitegrid", palette="viridis")
print("All libraries imported successfully!")
```

```python
# Load the dataset (make sure the CSV file is in the same folder as your notebook)
# Note: Check the exact filename in your folder; usually it is 'Global_Cybersecurity_Threats_2015-2024.csv'
df = pd.read_csv('Global_Cybersecurity_Threats_2015-2024.csv')
```

```python
# Strip any accidental leading/trailing spaces from column names
df.columns = df.columns.str.strip()

# Display dataset dimensions (rows, columns)
print(f"Dataset Shape: {df.shape[0]} incidents, {df.shape[1]} features\n")
```

```python
# Make a copy of the dataframe for ML modeling
df_ml = df.copy()
```

```python
# Automatically detect numerical columns for feature engineering
users_col = [col for col in df_ml.columns if 'Number of Affected' in col][0]

# --- IMPRESSIVE ADDITION: Domain-Specific Feature Engineering ---
# 1. Financial Loss per Affected User (measures severity per victim)
df_ml['Loss_Per_User'] = (df_ml[loss_col] * 1_000_000) / (df_ml[users_col] + 1)

# 2. Financial Impact Rate per Hour (measures how fast damage accumulates during downtime)
df_ml['Loss_Per_Hour'] = df_ml[loss_col] / (df_ml[res_col] + 1)

# Separate the target variable ('Attack Type') from input features
target_encoder = LabelEncoder()
y = target_encoder.fit_transform(df_ml['Attack Type'])

# Save class names for clear plotting later
class_names = target_encoder.classes_
print("Encoded Target Classes:", dict(enumerate(class_names)))

# Drop the target column from features (X)
X = df_ml.drop(columns=['Attack Type'])

# One-Hot Encode categorical columns (Country, Target Industry, Attack Source, Vulnerability, Defense)
X = pd.get_dummies(X, drop_first=True)

# Split the dataset: 80% Training, 20% Testing (stratified to preserve class balance)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Standardize numerical features (crucial for SVM, KNN, and Logistic Regression)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"\nTraining Set Shape: {X_train_scaled.shape}")
print(f"Testing Set Shape:  {X_test_scaled.shape}")
```

```python
# Initialize 6 diverse classifiers to compare
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "K-Nearest Neighbors": KNeighborsClassifier(n_neighbors=7),
    "Support Vector Machine": SVC(kernel='rbf', random_state=42),
    "Decision Tree": DecisionTreeClassifier(max_depth=10, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=150, random_state=42)
}

# List to store performance metrics for each model
results = []

# Loop through each model, train it, predict on test data, and compute all metrics
for name, model in models.items():
    # Train the model
    model.fit(X_train_scaled, y_train)
    
    # Generate predictions on unseen test data
    y_pred = model.predict(X_test_scaled)
    
    # Calculate the 4 scalar metrics (weighted average handles multi-class labels cleanly)
    acc = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred, average='weighted')
    rec = recall_score(y_test, y_pred, average='weighted')
    prec = precision_score(y_test, y_pred, average='weighted')
    
    results.append({
        "Model": name,
        "Accuracy": round(acc * 100, 2),
        "F1 Score": round(f1 * 100, 2),
        "Recall": round(rec * 100, 2),
        "Precision": round(prec * 100, 2)
    })

# Convert results into a sorted comparison DataFrame
results_df = pd.DataFrame(results).sort_values(by="F1 Score", ascending=False).reset_index(drop=True)
print("=== Model Performance Comparison (%) ===")
results_df
```

```python
# 1. Plot Grouped Bar Chart comparing all models across Accuracy, F1, Recall, and Precision
results_melted = results_df.melt(id_vars="Model", var_name="Metric", value_name="Score (%)")

plt.figure(figsize=(14, 6))
sns.barplot(data=results_melted, x="Model", y="Score (%)", hue="Metric")
plt.title("Cyber Threat Classification — Machine Learning Model Comparison", fontsize=14, fontweight='bold')
plt.ylim(0, 100)
plt.xticks(rotation=20)
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()

# 2. Generate Detailed Report & Confusion Matrix for the Best Performing Model
best_model_name = results_df.iloc[0]["Model"]
best_model = models[best_model_name]
y_pred_best = best_model.predict(X_test_scaled)

print(f"\nDetailed Classification Report for Top Model ({best_model_name}):\n")
print(classification_report(y_test, y_pred_best, target_names=class_names))

# Plot Confusion Matrix Heatmap
cm = confusion_matrix(y_test, y_pred_best)
plt.figure(figsize=(9, 7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.title(f"Confusion Matrix — {best_model_name}", fontsize=14, fontweight='bold')
plt.xlabel("Predicted Attack Type", fontsize=12)
plt.ylabel("Actual Attack Type", fontsize=12)
plt.xticks(rotation=30)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Extract feature importances from Random Forest to explain WHICH factors drive cyber attacks
rf_model = models["Random Forest"]
importances = pd.Series(rf_model.feature_importances_, index=X.columns)

# Plot the Top 12 most influential cybersecurity indicators
plt.figure(figsize=(10, 6))
importances.nlargest(12).sort_values().plot(kind='barh', color='#2b5c8f')
plt.title("Top 12 Indicators Driving Cyber Attack Classification", fontsize=13, fontweight='bold')
plt.xlabel("Relative Feature Importance Score")
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# EXPERIMENT 2: AUTOMATED SOC INCIDENT SEVERITY TRIAGE (BINARY CLASSIFICATION)
# ==============================================================================

# Create a fresh copy of the cleaned dataframe
df_sev = df.copy()

# Automatically grab the exact column names for Loss, Resolution Time, and Affected Users
loss_col = [col for col in df_sev.columns if 'Financial Loss' in col][0]
res_col = [col for col in df_sev.columns if 'Incident Resolution' in col][0]
users_col = [col for col in df_sev.columns if 'Number of Affected' in col][0]

# Define SOC Severity Policy:
# Critical (1) if Financial Loss is in the top 25% (75th percentile)
# OR if both Financial Loss and Resolution Time exceed their respective medians.
loss_75 = df_sev[loss_col].quantile(0.75)
loss_med = df_sev[loss_col].median()
res_med = df_sev[res_col].median()

df_sev['Severity_Label'] = np.where(
    (df_sev[loss_col] >= loss_75) | ((df_sev[loss_col] > loss_med) & (df_sev[res_col] > res_med)),
    1,  # 1 = Critical Incident (Immediate Tier-3 SOC Escalation)
    0   # 0 = Standard Incident (Routine Tier-1/Tier-2 Handling)
)

# Verify class balance between Standard (0) and Critical (1) incidents
print("Incident Severity Distribution:")
print(df_sev['Severity_Label'].value_counts().rename({0: 'Standard (0)', 1: 'Critical (1)'}))

# Separate features (X_sev) and binary target (y_sev)
# Drop loss_col and res_col to prevent target leakage from the threshold rule
y_sev = df_sev['Severity_Label']
X_sev = df_sev.drop(columns=['Severity_Label', loss_col, res_col])

# One-Hot Encode all categorical variables (including Attack Type)
X_sev = pd.get_dummies(X_sev, drop_first=True)

# Split into 80% Training and 20% Testing sets (stratified to maintain exact class ratio)
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_sev, y_sev, test_size=0.20, random_state=42, stratify=y_sev
)

# Scale numerical features strictly on training data to avoid data leakage
scaler_sev = StandardScaler()
X_train_s_scaled = scaler_sev.fit_transform(X_train_s)
X_test_s_scaled = scaler_sev.transform(X_test_s)

print(f"\nExperiment 2 Training Shape: {X_train_s_scaled.shape}")
print(f"Experiment 2 Testing Shape:  {X_test_s_scaled.shape}")
```

```python
# Re-initialize the same 6 models for a fair side-by-side comparison with Experiment 1
sev_models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "K-Nearest Neighbors": KNeighborsClassifier(n_neighbors=7),
    "Support Vector Machine": SVC(kernel='rbf', random_state=42),
    "Decision Tree": DecisionTreeClassifier(max_depth=6, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=150, random_state=42)
}

sev_results = []

# Train and evaluate each model across all 4 scalar metrics
for name, model in sev_models.items():
    model.fit(X_train_s_scaled, y_train_s)
    y_pred_s = model.predict(X_test_s_scaled)
    
    sev_results.append({
        "Model": name,
        "Accuracy": round(accuracy_score(y_test_s, y_pred_s) * 100, 2),
        "F1 Score": round(f1_score(y_test_s, y_pred_s) * 100, 2),
        "Recall": round(recall_score(y_test_s, y_pred_s) * 100, 2),
        "Precision": round(precision_score(y_test_s, y_pred_s) * 100, 2)
    })

# Display sorted performance table
sev_results_df = pd.DataFrame(sev_results).sort_values(by="F1 Score", ascending=False).reset_index(drop=True)
print("=== Experiment 2: Incident Severity Triage Performance (%) ===")
sev_results_df
```

```python
# 1. Grouped Bar Chart comparing model performance on Severity Triage
sev_melted = sev_results_df.melt(id_vars="Model", var_name="Metric", value_name="Score (%)")

plt.figure(figsize=(14, 6))
sns.barplot(data=sev_melted, x="Model", y="Score (%)", hue="Metric", palette="mako")
plt.title("Experiment 2: SOC Incident Severity Classification — Model Comparison", fontsize=14, fontweight='bold')
plt.ylim(0, 105)
plt.xticks(rotation=20)
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()

# 2. Classification Report & Confusion Matrix for the #1 Model
best_sev_name = sev_results_df.iloc[0]["Model"]
best_sev_model = sev_models[best_sev_name]
y_pred_sev_best = best_sev_model.predict(X_test_s_scaled)

print(f"\nDetailed Classification Report — {best_sev_name}:\n")
print(classification_report(y_test_s, y_pred_sev_best, target_names=['Standard (0)', 'Critical (1)']))

# Plot Confusion Matrix Heatmap
cm_sev = confusion_matrix(y_test_s, y_pred_sev_best)
plt.figure(figsize=(7, 5))
sns.heatmap(
    cm_sev, annot=True, fmt='d', cmap='Greens',
    xticklabels=['Standard (0)', 'Critical (1)'],
    yticklabels=['Standard (0)', 'Critical (1)']
)
plt.title(f"Severity Triage Confusion Matrix — {best_sev_name}", fontsize=13, fontweight='bold')
plt.xlabel("Predicted Severity Tier", fontsize=11)
plt.ylabel("Actual Severity Tier", fontsize=11)
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# 1. Yearly Financial Loss Trend (2015 - 2024)
yearly_loss = df.groupby('Year')['Financial Loss (in Million $)'].sum().reset_index()
sns.lineplot(
    data=yearly_loss, x='Year', y='Financial Loss (in Million $)', 
    marker='o', linewidth=2.5, color='#1f77b4', ax=axes[0]
)
axes[0].set_title("Total Global Financial Loss by Year (2015–2024)", fontweight='bold')
axes[0].set_ylabel("Total Loss ($ Millions)")
axes[0].grid(True, linestyle='--', alpha=0.5)

# 2. Top 10 Countries by Cumulative Financial Loss
top_countries = (
    df.groupby('Country')['Financial Loss (in Million $)']
    .sum()
    .nlargest(10)
    .sort_values()
)
top_countries.plot(kind='barh', color='#2ca02c', ax=axes[1])
axes[1].set_title("Top 10 Countries by Cumulative Financial Loss", fontweight='bold')
axes[1].set_xlabel("Total Loss ($ Millions)")
axes[1].set_ylabel("")

plt.tight_layout()
plt.show()
```
